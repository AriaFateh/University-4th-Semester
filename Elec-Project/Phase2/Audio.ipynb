```python
import soundfile as sf

# Name of your file (put the file in the same directory as this code if failed change the following string to the audios path)
input_file = r"input.wav"
output_file = r"audio.wav"
data, samplerate = sf.read(input_file)
sf.write(output_file, data, samplerate, subtype='PCM_16')

print(f"Corrected file saved to: {output_file}")
```
